### C. CATEGORY-AVERAGE VISUALIZATION

```python
import matplotlib.pyplot as plt
```

```python
track_mean = df.groupby("Track")["Average"].mean()
gender_mean = df.groupby("Gender")["Average"].mean()
hometown_mean = df.groupby("Hometown")["Average"].mean()
```

```python
print("Mean Average by Track:")
display(track_mean.reset_index())

print("Mean Average by Gender:")
display(gender_mean.reset_index())

print("Mean Average by Hometown:")
display(hometown_mean.reset_index())
```

```python

plt.figure(figsize=(18, 5))

# 1. Mean Average by Track
plt.subplot(1, 3, 1)
plt.bar(track_mean.index, track_mean.values)
plt.title("Mean Average by Track")
plt.xlabel("Track")
plt.ylabel("Mean Average")
plt.xticks(rotation=45)

# 2. Mean Average by Gender
plt.subplot(1, 3, 2)
plt.bar(gender_mean.index, gender_mean.values)
plt.title("Mean Average by Gender")
plt.xlabel("Gender")
plt.ylabel("Mean Average")

# 3. Mean Average by Hometown
plt.subplot(1, 3, 3)
plt.bar(hometown_mean.index, hometown_mean.values)
plt.title("Mean Average by Hometown")
plt.xlabel("Hometown")
plt.ylabel("Mean Average")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Based on the graph,

1. **Communication** has the highest sample mean Average among the Track categories.
2. **Male** has the highest sample mean Average among the Gender categories.
3. **Luzon** has the highest sample mean Average among the Hometown categories.
